```python
import urllib.request

url = "https://raw.githubusercontent.com/Khalid2572/Students-Performance-dataset/main/StudentsPerformance.csv"

Path("/content/data").mkdir(parents=True, exist_ok=True)

urllib.request.urlretrieve(
    url,
    "/content/data/StudentsPerformance.csv"
)

print("Датасет скачан")
```

```python
df = pd.read_csv("/content/data/StudentsPerformance.csv")

print("Данные успешно загружены")
print("Размер таблицы:", df.shape)

df.head()
```

```python
df.dtypes
```

```python
print("Количество пропусков по столбцам:")
print(df.isnull().sum())

print("\nОбщее количество пропусков:")
print(df.isnull().sum().sum())
```

```python
df.describe()
```

```python
df.describe(include="all")
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["math score"], bins=15, edgecolor="black")

plt.title("Распределение баллов по математике")
plt.xlabel("Баллы")
plt.ylabel("Количество учащихся")

plt.show()
```

```python
gender_counts = df["gender"].value_counts()

plt.figure(figsize=(7, 5))

gender_counts.plot(kind="bar", edgecolor="black")

plt.title("Количество учащихся по полу")
plt.xlabel("Категория")
plt.ylabel("Количество учащихся")
plt.xticks(rotation=0)

plt.show()
```

```python
print("Количество учащихся по полу:")
print(df["gender"].value_counts())

print("\nСредние баллы по математике в зависимости от подготовки:")
print(df.groupby("test preparation course")["math score"].mean().round(2))

print("\nСвязь между баллами по чтению и письму:")
print(df["reading score"].corr(df["writing score"]).round(3))
```
