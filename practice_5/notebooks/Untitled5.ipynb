## 2. Визуализация данных

### Распределение баллов по математике

```python
plt.figure(figsize=(8, 5))
plt.hist(df["math score"], bins=10)
plt.title("Распределение баллов по математике")
plt.xlabel("Баллы")
plt.ylabel("Количество студентов")
plt.show()
```

### Соотношение студентов по полу

```python
df["gender"].value_counts().plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Соотношение студентов по полу")
plt.xlabel("Пол")
plt.ylabel("Количество студентов")
plt.show()
```

```python
df.groupby("test preparation course")[["math score", "reading score", "writing score"]].mean()
```

```python
df.groupby("gender")[["math score", "reading score", "writing score"]].mean()
```

```python
df.groupby("lunch")[["math score", "reading score", "writing score"]].mean()
```

## 3. Основные наблюдения

1. Студенты, прошедшие подготовительный курс, в среднем показывают более высокие результаты по математике, чтению и письму по сравнению со студентами, которые его не проходили.

2. Результаты по чтению и письму имеют заметную связь с полом студентов: в среднем девушки показывают более высокие результаты по этим двум дисциплинам, тогда как по математике различие менее выражено.

3. Тип питания связан с результатами обучения: студенты со стандартным типом питания в среднем имеют более высокие результаты по всем трём предметам по сравнению со студентами с льготным типом питания.
